```python
print("=== 第一組測試：4人，折價券 100 元 ===")
# 1. 原始資料：保留文字資料，另外建立整理後的變數
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 整理聚餐名稱：去除頭尾空白並統一轉成小寫
trimmed_name = event_name.strip()
clean_event_name = trimmed_name.lower()

# 3. 清理餐費文字，再轉換餐費與人數的型別
cleaned_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_cost = float(cleaned_meal_text)
people_count = int(people)

# 4. 分步計算費用：先按原始餐費計算服務費，再扣除折價券
service_fee = meal_cost * service_rate
total_amount = meal_cost + service_fee - discount
cost_per_person = total_amount / people_count

# 5. 判斷是否超過預算：比較結果直接保存為 bool（等於預算時不算超過）
is_over_budget = cost_per_person > budget_per_person

# 6. 輸出結果
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_cost:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


print("=== 第二組測試：5人，折價券 100 元 ===")
# 1. 修改變數並重新計算（只將人數改成 "5"，折價券維持 100 元）
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "5"
discount = 100

# 2. 整理聚餐名稱
trimmed_name = event_name.strip()
clean_event_name = trimmed_name.lower()

# 3. 清理餐費與人數並轉換型別
cleaned_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_cost = float(cleaned_meal_text)
people_count = int(people)

# 4. 分步計算費用
service_fee = meal_cost * service_rate
total_amount = meal_cost + service_fee - discount
cost_per_person = total_amount / people_count

# 5. 判斷是否超過預算
is_over_budget = cost_per_person > budget_per_person

# 6. 輸出結果
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_cost:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


print("=== 第三組測試：4人，折價券 220 元 ===")
# 1. 修改變數並重新計算（人數恢復為 "4"，只將折價券改成 220 元）
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
discount = 220

# 2. 整理聚餐名稱
trimmed_name = event_name.strip()
clean_event_name = trimmed_name.lower()

# 3. 清理餐費與人數並轉換型別
cleaned_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_cost = float(cleaned_meal_text)
people_count = int(people)

# 4. 分步計算費用
service_fee = meal_cost * service_rate
total_amount = meal_cost + service_fee - discount
cost_per_person = total_amount / people_count

# 5. 判斷是否超過預算
is_over_budget = cost_per_person > budget_per_person

# 6. 輸出結果
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_cost:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
